```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import KFold
from sklearn.metrics import mean_absolute_error, mean_squared_error
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Input, Dropout
from tensorflow.keras.regularizers import l2
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
import matplotlib.pyplot as plt
import itertools

print("Num GPUs Available: ", len(tf.config.list_physical_devices('GPU')))

# Load and Preprocess Data
df = pd.read_csv(r"C:\Users\sathw\Desktop\NARL12117039\SN_ms_tot_V2.0_interpolated (1).csv", header=None)
df.columns = ["Year", "Month", "Decimal_Date", "Smoothed_Sunspot_Number", "Standard_Deviation", "Observations", "Definitive_Indicator"]

cycle_length = 132
amplitude = 50
phase_shift = np.pi / 2 - 2 * np.pi * 31 / 132
df["Cycle_Position"] = ((df["Year"] - 1755) * 12 + df["Month"]) % cycle_length
df["Baseline"] = amplitude * (np.sin(2 * np.pi * df["Cycle_Position"] / cycle_length + phase_shift) + 1)
df["Residual"] = df["Smoothed_Sunspot_Number"] - df["Baseline"]

print("Residual Statistics:")
print(df["Residual"].describe())
print("Last Date in Dataset:", pd.to_datetime(f"{int(df['Year'].iloc[-1])}-{int(df['Month'].iloc[-1]):02d}"))

features = df[["Year", "Month", "Cycle_Position"]].values
target = df["Residual"].values.reshape(-1, 1)

scaler_features = MinMaxScaler()
scaler_target = MinMaxScaler()
features_scaled = scaler_features.fit_transform(features)
target_scaled = scaler_target.fit_transform(target)

# Create Sequences
def create_sequences(features, target, seq_length):
    X, y = [], []
    for i in range(len(target) - seq_length):
        X.append(features[i:i + seq_length])
        y.append(target[i + seq_length])
    return np.array(X), np.array(y)

seq_length = 24
X, y = create_sequences(features_scaled, target_scaled, seq_length)

# Define LSTM Model
def build_model(seq_length, n_features, lstm_units, dropout_rate, learning_rate, l2_lambda=0.01):
    model = Sequential([
        Input(shape=(seq_length, n_features)),
        LSTM(lstm_units, activation='tanh', return_sequences=True, kernel_regularizer=l2(l2_lambda)),
        LSTM(lstm_units // 2, activation='tanh', kernel_regularizer=l2(l2_lambda)),
        Dropout(dropout_rate),
        Dense(1, kernel_regularizer=l2(l2_lambda))
    ])
    optimizer = tf.keras.optimizers.Adam(learning_rate=learning_rate, clipnorm=1.0)
    model.compile(optimizer=optimizer, loss='mse', metrics=['mae'])
    return model

# Hyperparameter Tuning with K-Fold
param_grid = {
    'lstm_units': [32, 64],
    'dropout_rate': [0.2, 0.3],
    'learning_rate': [0.0005, 0.001]
}

best_val_loss = float('inf')
best_params = None
k = 3
kf = KFold(n_splits=k, shuffle=False)

for lstm_units, dropout_rate, learning_rate in itertools.product(param_grid['lstm_units'], param_grid['dropout_rate'], param_grid['learning_rate']):
    print(f"Testing params: lstm_units={lstm_units}, dropout_rate={dropout_rate}, learning_rate={learning_rate}")
    fold_scores = []
    for fold, (train_idx, val_idx) in enumerate(kf.split(X)):
        print(f"Training Fold {fold + 1}/{k}")
        X_train, X_val = X[train_idx], X[val_idx]
        y_train, y_val = y[train_idx], y[val_idx]
        model = build_model(seq_length, X.shape[2], lstm_units, dropout_rate, learning_rate)
        model.fit(X_train, y_train, epochs=5, batch_size=64, validation_data=(X_val, y_val), verbose=0)
        val_metrics = model.evaluate(X_val, y_val, verbose=0)
        val_loss = val_metrics[0]
        if np.isnan(val_loss) or np.isinf(val_loss):
            print(f"Fold {fold + 1} encountered NaN/Inf loss, skipping...")
            continue
        fold_scores.append(val_loss)
        print(f"Fold {fold + 1} Validation Loss: {val_loss:.4f}")
    avg_val_loss = np.mean(fold_scores)
    print(f"Average Validation Loss: {avg_val_loss:.4f}")
    if avg_val_loss < best_val_loss:
        best_val_loss = avg_val_loss
        best_params = {'lstm_units': lstm_units, 'dropout_rate': dropout_rate, 'learning_rate': learning_rate}

print("Best Parameters:", best_params)
print("Best Average Validation Loss:", best_val_loss)

# Train Final Model
final_model = build_model(seq_length, X.shape[2], best_params['lstm_units'], best_params['dropout_rate'], best_params['learning_rate'])

early_stopping = EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True, verbose=1)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=3, min_lr=0.00001, verbose=1)

history = final_model.fit(
    X, y,
    epochs=100,
    batch_size=64,
    validation_split=0.2,
    callbacks=[early_stopping, reduce_lr],
    verbose=1
)

# Evaluate Model on Validation Set
val_split = 0.2
val_idx = int(len(X) * (1 - val_split))
X_train, X_val = X[:val_idx], X[val_idx:]
y_train, y_val = y[:val_idx], y[val_idx:]
y_val_pred_scaled = final_model.predict(X_val, verbose=0)
y_val_pred = scaler_target.inverse_transform(y_val_pred_scaled)
y_val_true = scaler_target.inverse_transform(y_val)

mae = mean_absolute_error(y_val_true, y_val_pred)
mse = mean_squared_error(y_val_true, y_val_pred)
rmse = np.sqrt(mse)

print("\nEvaluation Metrics on Validation Set:")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")

# Predict Future up to Dec 2030
def predict_future(model, last_sequence, n_future, scaler_features, scaler_target, last_year_month_cycle):
    future_preds = []
    current_seq = last_sequence.copy()
    last_year, last_month, last_cycle_pos = last_year_month_cycle
    for _ in range(n_future):
        pred = model.predict(current_seq[np.newaxis, :, :], verbose=0)
        future_preds.append(pred[0, 0])
        next_features = current_seq[1:, :]
        new_month = (last_month + 1) % 12 or 12
        new_year = last_year + (last_month + 1) // 12
        new_cycle_pos = (last_cycle_pos + 1) % cycle_length
        new_feature = np.array([[new_year, new_month, new_cycle_pos]])
        new_feature_scaled = scaler_features.transform(new_feature)
        current_seq = np.vstack([next_features, new_feature_scaled])
        last_year, last_month, last_cycle_pos = new_year, new_month, new_cycle_pos
    future_preds = np.array(future_preds).reshape(-1, 1)
    return scaler_target.inverse_transform(future_preds)

last_sequence = X[-1]
last_year_month_cycle = scaler_features.inverse_transform(last_sequence[-1].reshape(1, -1))[0]
n_future = 72  # Jan 2025 to Dec 2030
future_residuals = predict_future(final_model, last_sequence, n_future, scaler_features, scaler_target, last_year_month_cycle)

future_cycle_pos = np.arange(df["Cycle_Position"].iloc[-1] + 1, df["Cycle_Position"].iloc[-1] + 73) % cycle_length
future_baseline = amplitude * (np.sin(2 * np.pi * future_cycle_pos / cycle_length + phase_shift) + 1)
future_predictions = np.clip(future_residuals + future_baseline.reshape(-1, 1), 0, None)

# Diagnostics and Visualization
print("Future Residuals Stats:", future_residuals.min(), future_residuals.max())
print("Future Predictions Statistics:")
print(f"Min: {future_predictions.min():.2f}, Max: {future_predictions.max():.2f}")
print(f"Mean: {future_predictions.mean():.2f}, Std: {future_predictions.std():.2f}")
print("First 12 predictions (1 year):", future_predictions[:12].flatten())

plt.figure(figsize=(14, 7))
historical_dates = pd.date_range(start="1749-01", periods=len(df), freq="MS")[-len(y):]
future_dates = pd.date_range(start="2025-01", periods=n_future, freq="MS")
plt.plot(historical_dates, df["Smoothed_Sunspot_Number"].values[-len(y):], label='Historical Sunspots')
plt.plot(future_dates, future_predictions, label='Predicted Sunspots (Cycle 25)')
plt.plot(future_dates, future_baseline, label='Baseline', linestyle='--')
plt.axvline(x=pd.to_datetime("2025-01-01"), color='gray', linestyle='--', label='Prediction Start')
plt.xlabel('Year')
plt.ylabel('Smoothed Sunspot Number')
plt.title('Solar Cycle 25 Prediction LSTM')
plt.legend()
plt.grid(True)
plt.show()

plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training and Validation Loss')
plt.legend()
plt.show()

# Save Predictions
start_date = "2025-01"
dates_cycle_25 = pd.date_range(start=start_date, periods=n_future, freq="MS")
pd.DataFrame({
    "Year": dates_cycle_25.year,
    "Month": dates_cycle_25.month,
    "Predicted_Sunspot_Number": future_predictions.flatten()
}).to_csv("solar_cycle_25_lstm_predictions_dec2030.csv", index=False)

print("✅ LSTM Predictions for Solar Cycle 25 saved successfully!")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Define cycle starts and prediction period
cycle_23_start = pd.to_datetime("1996-05-01")  # Cycle 23 minimum
cycle_24_start = pd.to_datetime("2008-12-01")  # Cycle 24 minimum
cycle_25_start = pd.to_datetime("2019-12-01")  # Cycle 25 minimum
cycle_length_months = 132
last_date = pd.to_datetime(f"{int(df['Year'].iloc[-1])}-{int(df['Month'].iloc[-1]):02d}-01")

# Historical dates (aligned with df)
historical_dates = pd.date_range(start="1749-01-01", end=last_date, freq="MS")
if len(historical_dates) != len(df):
    raise ValueError(f"Length mismatch: historical_dates ({len(historical_dates)}) vs df ({len(df)})")

# Extract Cycles 23 and 24 (extend Cycle 23 to meet Cycle 24)
cycle_23_mask = (historical_dates >= cycle_23_start) & (historical_dates < cycle_24_start)  # End at Cycle 24 start
cycle_24_mask = (historical_dates >= cycle_24_start) & (historical_dates < cycle_24_start + pd.offsets.MonthEnd(cycle_length_months))
cycle_23_data = df["Smoothed_Sunspot_Number"][cycle_23_mask].values
cycle_24_data = df["Smoothed_Sunspot_Number"][cycle_24_mask].values
cycle_23_dates = historical_dates[cycle_23_mask]
cycle_24_dates = historical_dates[cycle_24_mask]

# Verify continuity between Cycle 23 and 24
if cycle_23_dates[-1] + pd.offsets.MonthBegin(1) != cycle_24_dates[0]:
    print(f"Warning: Gap between Cycle 23 end ({cycle_23_dates[-1]}) and Cycle 24 start ({cycle_24_dates[0]})")
else:
    print("Cycle 23 and 24 dates are continuous")

# Verify lengths
if len(cycle_23_data) != len(cycle_23_dates):
    print(f"Warning: Cycle 23 data ({len(cycle_23_data)}) and dates ({len(cycle_23_dates)}) mismatch")
if len(cycle_24_data) != len(cycle_24_dates):
    print(f"Warning: Cycle 24 data ({len(cycle_24_data)}) and dates ({len(cycle_24_dates)}) mismatch")

# Extract Cycle 25 historical data (Dec 2019 to last_date) + predictions (Jan 2025–Dec 2030)
cycle_25_hist_mask = (historical_dates >= cycle_25_start) & (historical_dates <= last_date)
cycle_25_hist_data = df["Smoothed_Sunspot_Number"][cycle_25_hist_mask].values
cycle_25_hist_dates = historical_dates[cycle_25_hist_mask]
cycle_25_pred_dates = pd.date_range(start=last_date + pd.offsets.MonthBegin(1), periods=72, freq="MS")

# Handle future_predictions (assume future_preds from LSTM code)
try:
    cycle_25_full_data = np.concatenate([cycle_25_hist_data, future_predictions.flatten()])  # Original variable
except NameError:
    cycle_25_full_data = np.concatenate([cycle_25_hist_data, future_preds.flatten()])  # Use future_preds from LSTM

cycle_25_full_dates = np.concatenate([cycle_25_hist_dates, cycle_25_pred_dates])

# Verify Cycle 25 lengths
if len(cycle_25_full_data) != len(cycle_25_full_dates):
    raise ValueError(f"Cycle 25 data ({len(cycle_25_full_data)}) and dates ({len(cycle_25_full_dates)}) mismatch")

# Smooth Cycle 25
def smooth_sunspots(data, window=13):
    pad = (window - 1) // 2
    padded_data = np.pad(data, (pad, pad), mode='edge')
    smoothed = np.convolve(padded_data, np.ones(window)/window, mode='valid')
    if len(smoothed) != len(data):
        smoothed = smoothed[:len(data)]  # Truncate to match original length
    return smoothed

cycle_25_smoothed = smooth_sunspots(cycle_25_full_data)

# Plot
plt.figure(figsize=(14, 7))
plt.plot(cycle_23_dates, cycle_23_data, label='Cycle 23 (1996-2007)', color='blue')
plt.plot(cycle_24_dates, cycle_24_data, label='Cycle 24 (2008-2019)', color='green')
plt.plot(cycle_25_full_dates, cycle_25_smoothed, label='Cycle 25 (2019-2030, Hist + Pred)', color='orange')
plt.axvline(x=last_date, color='gray', linestyle='--', label='Prediction Start')
plt.xlabel('Year')
plt.ylabel('Smoothed Sunspot Number')
plt.title('Solar Cycles 23, 24, and 25 (Historical and Predicted)')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
