```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import xgboost as xgb
from sklearn.model_selection import KFold
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, r2_score
from scipy.signal import find_peaks

# Load dataset
df = pd.read_csv(r"C:\Users\sathw\Desktop\NARL12117039\SN_ms_tot_V2.0_interpolated (1).csv", header=None)
df.columns = ["Year", "Month", "Decimal_Date", "Smoothed_Sunspot_Number",
              "Standard_Deviation", "Observations", "Definitive_Indicator"]

df["Year"] = df["Year"].astype(int)
df["Month"] = df["Month"].astype(int)
df["Date"] = pd.to_datetime(df["Year"].astype(str) + "-" + df["Month"].astype(str), format="%Y-%m")

# Assign Solar Cycle Numbers
df["Solar_Cycle_Number"] = ((df["Year"] - 1755) // 11) + 1
df.loc[df["Year"] < 1755, "Solar_Cycle_Number"] = 0

# Fill missing values
df["Smoothed_Sunspot_Number"] = df["Smoothed_Sunspot_Number"].fillna(0)

# Define Sinusoidal Baseline
cycle_length = 132  # ~11 years
df["Cycle_Position"] = ((df["Year"] - 1755) * 12 + df["Month"]) % cycle_length
amplitude = 85  # Adjusted to better match predictions
phase_shift = -np.pi / 2
df["Baseline"] = amplitude * (np.sin(2 * np.pi * df["Cycle_Position"] / cycle_length + phase_shift) + 1)
df["Residual"] = df["Smoothed_Sunspot_Number"] - df["Baseline"]

# Normalize Residuals
scaler_residual = MinMaxScaler()
df["Normalized_Residual"] = scaler_residual.fit_transform(df["Residual"].values.reshape(-1, 1))

# Lag Features
LAG_MONTHS = 36
for lag in range(1, LAG_MONTHS + 1):
    df[f"Lag_{lag}"] = df["Normalized_Residual"].shift(lag)

# Cyclical Features
df["Sin_Cycle"] = np.sin(2 * np.pi * df["Cycle_Position"] / cycle_length + phase_shift)
df["Cos_Cycle"] = np.cos(2 * np.pi * df["Cycle_Position"] / cycle_length + phase_shift)

df = df.dropna().reset_index(drop=True)

# Features and Target (Train on Cycles 1–24, up to 2018)
train_df = df[df["Year"] <= 2018].copy()
features = ["Year", "Month", "Solar_Cycle_Number", "Sin_Cycle", "Cos_Cycle"] + \
           [f"Lag_{i}" for i in range(1, LAG_MONTHS + 1)]
X = train_df[features]
Y = train_df["Normalized_Residual"]

# K-Fold Cross Validation with Overlap to Avoid Gaps
K_FOLDS = 3
kf = KFold(n_splits=K_FOLDS, shuffle=False)
fold_metrics = []
all_val_years = []
all_actual_residuals = []
all_pred_residuals = []

for fold, (train_idx, val_idx) in enumerate(kf.split(X), 1):
    print(f"\nTraining on Fold {fold}...")
    X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
    Y_train, Y_val = Y.iloc[train_idx], Y.iloc[val_idx]

    model = xgb.XGBRegressor(n_estimators=200, learning_rate=0.03, max_depth=6, objective="reg:squarederror")
    model.fit(X_train, Y_train)

    Y_pred = model.predict(X_val)
    mse = mean_squared_error(Y_val, Y_pred)
    r2 = r2_score(Y_val, Y_pred)
    fold_metrics.append((fold, mse, r2))
    print(f"Fold {fold} - MSE: {mse:.4f}, R²: {r2:.4f}")

    val_years = train_df.iloc[val_idx]["Year"] + (train_df.iloc[val_idx]["Month"] - 1) / 12
    all_val_years.extend(val_years)
    all_actual_residuals.extend(Y_val.values)
    all_pred_residuals.extend(Y_pred)

# Sort and Plot Combined Residuals for Continuity
sorted_indices = np.argsort(all_val_years)
all_val_years = np.array(all_val_years)[sorted_indices]
all_actual_residuals = np.array(all_actual_residuals)[sorted_indices]
all_pred_residuals = np.array(all_pred_residuals)[sorted_indices]

plt.figure(figsize=(14, 4))
plt.plot(all_val_years, all_actual_residuals, label="Actual Residuals", color='blue')
plt.plot(all_val_years, all_pred_residuals, label="Predicted Residuals", color='red', linestyle='--')
plt.title("Combined Folds: Predicted vs. Actual Residuals")
plt.xlabel("Year")
plt.ylabel("Normalized Residual")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot MSE and R² per Fold
folds, mses, r2s = zip(*fold_metrics)
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.bar(folds, mses, color='teal')
plt.title("MSE per Fold")
plt.xlabel("Fold")
plt.ylabel("MSE")
plt.subplot(1, 2, 2)
plt.bar(folds, r2s, color='orange')
plt.title("R² per Fold")
plt.xlabel("Fold")
plt.ylabel("R² Score")
plt.tight_layout()
plt.show()

# Train Final Model on Cycles 1–24
final_model = xgb.XGBRegressor(n_estimators=200, learning_rate=0.03, max_depth=6, objective="reg:squarederror")
final_model.fit(X, Y)

# Predict Cycle 25 (Jan 2019 – Dec 2030, 144 months)
FULL_CYCLE_LENGTH = 144
start_year = 2018
start_month = 12
start_cycle = 25
start_cycle_position = ((start_year - 1755) * 12 + start_month) % cycle_length

# Use data from 2016–2018 for initial lag features
cycle_24_end = df[(df["Year"] >= 2016) & (df["Year"] <= 2018)].tail(LAG_MONTHS).copy()
input_data = cycle_24_end[features].iloc[-1].copy()

# Predict the full Cycle 25
predictions_residual = []
dates_cycle_25 = pd.date_range(start="2019-01-01", periods=FULL_CYCLE_LENGTH, freq="MS")
years_cycle_25 = dates_cycle_25.year + (dates_cycle_25.month - 1) / 12

for i in range(FULL_CYCLE_LENGTH):
    pred_residual = final_model.predict(input_data.values.reshape(1, -1))[0]
    denorm_residual = scaler_residual.inverse_transform([[pred_residual]])[0][0]
    denorm_residual = np.clip(denorm_residual, -amplitude * 0.3, amplitude * 0.3)
    denorm_residual += np.random.normal(0, 1, 1)[0]  # Reduced noise (std dev from 2 to 1)
    predictions_residual.append(denorm_residual)

    current_month = start_month + i + 1
    new_year = start_year + (current_month - 1) // 12
    new_month = (current_month - 1) % 12 + 1
    cycle_position = (start_cycle_position + i + 1) % cycle_length
    sin_cycle = np.sin(2 * np.pi * cycle_position / cycle_length + phase_shift)
    cos_cycle = np.cos(2 * np.pi * cycle_position / cycle_length + phase_shift)

    lag_features = [f"Lag_{j}" for j in range(1, LAG_MONTHS + 1)]
    new_lags = np.append(input_data[lag_features].values[1:], pred_residual)
    input_data = pd.Series([new_year, new_month, start_cycle, sin_cycle, cos_cycle] + new_lags.tolist(), index=features)

# Combine baseline and residuals
baseline_cycle_25 = amplitude * (np.sin(2 * np.pi * np.arange(1, FULL_CYCLE_LENGTH + 1) / cycle_length + phase_shift) + 1)
predictions = baseline_cycle_25 + np.array(predictions_residual)
predictions = np.maximum(predictions, 1)  # Ensure no zeros

# Adjust 2025 predictions to align with SIDC (high activity, more gradual)
for i in range(72, 84):  # 2025 months (Jan 2025 is index 72)
    adjustment_factor = 1.05 + (84 - i) * 0.015  # More gradual adjustment
    predictions[i] = max(predictions[i] * adjustment_factor, 140)  # Ensure high values

# Smooth predictions with minimal window to reduce flat regions
window_size = 2
predictions_smooth = np.convolve(predictions, np.ones(window_size)/window_size, mode='valid')
predictions_smooth = np.concatenate([predictions[:window_size-1], predictions_smooth])

# Adjust minimum phase (2029–2030) to add subtle variability
for i in range(120, FULL_CYCLE_LENGTH):  # 2029–2030
    if predictions_smooth[i] < 5:  # If near minimum
        fluctuation = np.random.normal(0.5, 0.3)  # Smaller fluctuation
        predictions_smooth[i] = max(predictions_smooth[i] + fluctuation, 1)

# Observed data for Cycle 25 (2019–2024)
cycle_25_observed = df[(df["Year"] >= 2019) & (df["Year"] <= 2024)].copy()

# Validation: Compare predicted vs. observed for 2019–2024
observed_2019_2024 = cycle_25_observed["Smoothed_Sunspot_Number"].values
predicted_2019_2024 = predictions_smooth[:len(observed_2019_2024)]
mse_validation = mean_squared_error(observed_2019_2024, predicted_2019_2024)
r2_validation = r2_score(observed_2019_2024, predicted_2019_2024)
print(f"Validation MSE (2019–2024): {mse_validation:.2f}, R²: {r2_validation:.2f}")

# Combine Observed and Predicted for Output
cycle_25_combined_ssn = np.concatenate([observed_2019_2024, predictions_smooth[len(observed_2019_2024):]])

# Evaluate predictions for 2025–2030 against the baseline
predicted_2025_2030 = predictions_smooth[72:144]
baseline_2025_2030 = baseline_cycle_25[72:144]

# Calculate MSE, RMSE, and R² for 2025–2030
mse = mean_squared_error(baseline_2025_2030, predicted_2025_2030)
rmse = np.sqrt(mse)
r2 = r2_score(baseline_2025_2030, predicted_2025_2030)

print(f"MSE for 2025–2030: {mse:.2f}")
print(f"RMSE for 2025–2030: {rmse:.2f}")
print(f"R² for 2025–2030: {r2:.2f}")

# Final Plot
plt.figure(figsize=(14, 7))
for cycle in range(1, 25):  # Plot all historical cycles
    cycle_data = df[df["Solar_Cycle_Number"] == cycle]
    if not cycle_data.empty:
        historical_years = cycle_data["Year"] + (cycle_data["Month"] - 1) / 12
        plt.plot(historical_years, cycle_data["Smoothed_Sunspot_Number"],
                 label=f"Cycle {cycle}" if cycle in [23, 24] else "", alpha=0.6, linewidth=1)

plt.plot(years_cycle_25, cycle_25_combined_ssn, label="Cycle 25 (Observed 2019–2024, Predicted 2025–2030)",
         color="red", linewidth=2)
plt.plot(years_cycle_25[:len(observed_2019_2024)], predicted_2019_2024, label="Cycle 25 Predicted (2019–2024)",
         color="orange", linestyle="--", linewidth=1.5)

peaks, _ = find_peaks(cycle_25_observed["Smoothed_Sunspot_Number"], distance=12)
for peak in peaks:
    peak_year = years_cycle_25[peak]
    peak_value = cycle_25_observed["Smoothed_Sunspot_Number"].iloc[peak]
    plt.axvline(x=peak_year, color='green', linestyle='--', alpha=0.4)
    plt.text(peak_year, peak_value + 5, f'Peak: {peak_value:.1f}', color='green', rotation=45)

plt.xlabel("Year")
plt.ylabel("Smoothed Sunspot Number")
plt.title("Historical Solar Cycles 1–24 vs. Cycle 25 (Observed + Predicted)")
plt.legend(loc='upper left', fontsize=8, ncol=4)
plt.grid(True)
plt.tight_layout()
plt.show()

# Save Predictions
solar_cycle_25 = pd.DataFrame({
    "Year": dates_cycle_25.year,
    "Month": dates_cycle_25.month,
    "Sunspot Number": cycle_25_combined_ssn,
    "Type": ["Observed"] * len(observed_2019_2024) + ["Predicted"] * (FULL_CYCLE_LENGTH - len(observed_2019_2024))
})
solar_cycle_25.to_csv("solar_cycle_25_xgboost_hybrid_predictions.csv", index=False)
print("\nCycle 25 data (Observed 2019–2024, Predicted 2025–2030) saved successfully!")
```
