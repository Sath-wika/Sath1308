```python
import pandas as pd
import matplotlib.pyplot as plt

# Load each prediction
df_tcn = pd.read_csv(r"C:\Users\sathw\Downloads\TCN_SolarCycle25_Predictions.csv")
df_lstm = pd.read_csv(r"C:\Users\sathw\Desktop\NARL12117039\solar_cycle_25_lstm_predictions_dec2030.csv")
df_xgb = pd.read_csv(r"C:\Users\sathw\Downloads\solar_cycle_25_xgboost_hybrid_predictions.csv")

# Load actual sunspot data
df_actual = pd.read_csv(r"C:\Users\sathw\Desktop\NARL12117039\SN_ms_tot_V2.0_interpolated (1).csv")

# Rename columns of actual data to meaningful names
df_actual.columns = ['Year', 'Month', 'Decimal_Year', 'Smoothed_Sunspot_Number', 'Unknown1', 'Unknown2', 'Unknown3']

# Fix column names for LSTM and XGBoost to match
df_lstm.rename(columns={'Predicted_Sunspot_Number': 'Predicted Sunspot Number'}, inplace=True)
df_xgb.rename(columns={'Sunspot Number': 'Predicted Sunspot Number'}, inplace=True)

# Print column info to confirm
print("TCN Columns:", df_tcn.columns)
print("LSTM Columns:", df_lstm.columns)
print("XGB Columns:", df_xgb.columns)
print("Actual Data Columns:", df_actual.columns)

# Plot all model predictions and actual data
plt.figure(figsize=(14, 6))

# Plot Actual Data (Smoothed Sunspot Number)
plt.plot(df_actual['Year'] + (df_actual['Month'] - 1) / 12, df_actual['Smoothed_Sunspot_Number'], 
         label="Actual Data", color='blue', linewidth=2)

# Plot TCN model predictions
plt.plot(df_tcn['Year'] + (df_tcn['Month'] - 1) / 12, df_tcn['Predicted Sunspot Number'], 
         label="TCN", linestyle='--', color='red')

# Plot LSTM model predictions
plt.plot(df_lstm['Year'] + (df_lstm['Month'] - 1) / 12, df_lstm['Predicted Sunspot Number'], 
         label="LSTM", linestyle=':', color='green')

# Plot XGBoost model predictions
plt.plot(df_xgb['Year'] + (df_xgb['Month'] - 1) / 12, df_xgb['Predicted Sunspot Number'], 
         label="XGBoost", linestyle='-.', color='purple')

# Customize the plot
plt.xlabel("Year")
plt.ylabel("Smoothed Sunspot Number")
plt.title("Comparison of Solar Cycle 25 Predictions vs Actual Data (2025–2030)")
plt.legend()
plt.grid(True)
plt.tight_layout()

# Show the plot
plt.show()
```
