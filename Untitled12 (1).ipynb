```python
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from torch.utils.data import DataLoader, TensorDataset
from sklearn.model_selection import KFold
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
from scipy.signal import savgol_filter
import random

# Set random seed for reproducibility
torch.manual_seed(42)
np.random.seed(42)
random.seed(42)

# Load Dataset
df = pd.read_csv(r"C:\Users\sathw\Desktop\NARL12117039\SN_ms_tot_V2.0_interpolated (1).csv")

# Rename columns correctly
df.columns = ["Year", "Month", "Decimal_Date", "Smoothed_Sunspot_Number", "Standard_Deviation", "Observations", "Definitive_Indicator"]

# Ensure proper data types
df['Year'] = df['Year'].astype(int)
df['Month'] = df['Month'].astype(int)

# Fix Solar Cycle Number (Explicit Cycle Encoding)
df['Solar_Cycle_Number'] = ((df['Year'] - 1755) // 11) + 1

# Add Sinusoidal Features for Periodicity
df['sin_month'] = np.sin(2 * np.pi * df['Month'] / 12)
df['cos_month'] = np.cos(2 * np.pi * df['Month'] / 12)
df['sin_cycle'] = np.sin(2 * np.pi * df['Solar_Cycle_Number'] / 11)
df['cos_cycle'] = np.cos(2 * np.pi * df['Solar_Cycle_Number'] / 11)
df['sin_year'] = np.sin(2 * np.pi * df['Year'] / 11)
df['cos_year'] = np.cos(2 * np.pi * df['Year'] / 11)

# Extract features
features = df[['Smoothed_Sunspot_Number', 'sin_month', 'cos_month', 'sin_cycle', 'cos_cycle', 'sin_year', 'cos_year']].values
solar_cycles = df['Solar_Cycle_Number'].values

# Normalize sunspot numbers
scaler = MinMaxScaler()
features[:, 0] = scaler.fit_transform(features[:, 0].reshape(-1, 1)).flatten()

# Function to Create Sequences
def create_sequences(data, cycle_labels, seq_length=132):
    sequences, labels, cycles = [], [], []
    for i in range(len(data) - seq_length):
        sequences.append(data[i:i+seq_length])
        labels.append(data[i+seq_length, 0])
        cycles.append(cycle_labels[i+seq_length])
    return np.array(sequences), np.array(labels), np.array(cycles)

# Prepare sequences
seq_length = 132
X, y, cycle_labels = create_sequences(features, solar_cycles, seq_length)
X_tensor = torch.tensor(X, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32)

# TCN Model
class TemporalConvNet(nn.Module):
    def __init__(self, input_size, output_size, num_channels, kernel_size=3):
        super(TemporalConvNet, self).__init__()
        layers = []
        for i in range(len(num_channels)):
            dilation_size = 2 ** i
            layers.append(nn.Conv1d(in_channels=input_size if i == 0 else num_channels[i-1],
                                    out_channels=num_channels[i],
                                    kernel_size=kernel_size,
                                    dilation=dilation_size,
                                    padding=(kernel_size - 1) * dilation_size))
            layers.append(nn.ReLU())
            layers.append(nn.Dropout(0.3))
        self.tcn = nn.Sequential(*layers)
        self.fc = nn.Linear(num_channels[-1], output_size)

    def forward(self, x):
        x = self.tcn(x)
        return self.fc(x[:, :, -1])

# Parameters
input_size = X.shape[2]
output_size = 1
num_channels = [32, 64, 128]
kernel_size = 3
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Custom Loss Function to Penalize Low Predictions
class CustomLoss(nn.Module):
    def __init__(self, alpha=0.1):
        super(CustomLoss, self).__init__()
        self.mse = nn.MSELoss()
        self.alpha = alpha

    def forward(self, output, target):
        mse_loss = self.mse(output, target)
        penalty = self.alpha * torch.mean(torch.clamp(-output, min=0) ** 2)
        return mse_loss + penalty

# Cross-validation
unique_cycles = np.unique(solar_cycles)
kf = KFold(n_splits=3, shuffle=False)

fold_results = []
lrs_all = []
best_model_state = None
best_val_r2 = 0
last_val_loader = None

for fold, (train_idx, val_idx) in enumerate(kf.split(unique_cycles)):
    print(f"\nFold {fold+1} Training...")

    train_mask = np.isin(cycle_labels, unique_cycles[train_idx])
    val_mask = np.isin(cycle_labels, unique_cycles[val_idx])

    X_train, X_val = X_tensor[train_mask], X_tensor[val_mask]
    y_train, y_val = y_tensor[train_mask], y_tensor[val_mask]

    train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=16, shuffle=True)
    val_loader = DataLoader(TensorDataset(X_val, y_val), batch_size=16, shuffle=False)

    model = TemporalConvNet(input_size, output_size, num_channels, kernel_size).to(device)
    optimizer = optim.Adam(model.parameters(), lr=0.0007, weight_decay=2e-4)
    scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)
    criterion = CustomLoss(alpha=0.1)

    num_epochs = 80
    best_val_loss = float('inf')
    best_fold_state = None
    patience = 15
    counter = 0
    lrs = []

    for epoch in range(num_epochs):
        model.train()
        train_loss = 0
        for xb, yb in train_loader:
            xb, yb = xb.permute(0, 2, 1).to(device), yb.to(device)
            optimizer.zero_grad()
            output = model(xb)
            loss = criterion(output.squeeze(), yb)
            loss.backward()
            optimizer.step()
            train_loss += loss.item()
            lrs.append(optimizer.param_groups[0]['lr'])

        model.eval()
        val_loss = 0
        y_true = []
        y_pred = []
        with torch.no_grad():
            for xb, yb in val_loader:
                xb, yb = xb.permute(0, 2, 1).to(device), yb.to(device)
                output = model(xb)
                loss = criterion(output.squeeze(), yb)
                val_loss += loss.item()
                y_true.extend(yb.cpu().numpy())
                y_pred.extend(output.squeeze().cpu().numpy())

        val_loss /= len(val_loader)
        val_r2 = r2_score(y_true, y_pred)
        print(f"Epoch {epoch+1}: Train Loss={train_loss / len(train_loader):.4f}, Val Loss={val_loss:.4f}, R²={val_r2:.4f}")

        scheduler.step(val_loss)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_fold_state = model.state_dict()
            counter = 0
            if val_r2 > best_val_r2:
                best_val_r2 = val_r2
                best_model_state = model.state_dict()
        else:
            counter += 1
            if counter >= patience:
                print(f"Early stopping at epoch {epoch+1}")
                model.load_state_dict(best_fold_state)
                break

    y_true = scaler.inverse_transform(np.array(y_true).reshape(-1, 1)).flatten()
    y_pred = scaler.inverse_transform(np.array(y_pred).reshape(-1, 1)).flatten()
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f"Fold {fold+1} MAE: {mae:.4f}, RMSE: {rmse:.4f}, R²: {r2:.4f}")
    fold_results.append({'MAE': mae, 'RMSE': rmse, 'R²': r2})
    lrs_all.extend(lrs)
    last_val_loader = val_loader  # Save the last fold's validation loader

# Plot learning rate
plt.figure(figsize=(10, 3))
plt.plot(lrs_all)
plt.title("Learning Rate Schedule")
plt.xlabel("Iteration")
plt.ylabel("Learning Rate")
plt.grid(True)
plt.show()

# Retrain on full data
print("\nRetraining final model on all data...")
full_loader = DataLoader(TensorDataset(X_tensor, y_tensor), batch_size=16, shuffle=True)
final_model = TemporalConvNet(input_size, output_size, num_channels, kernel_size).to(device)
if best_model_state:
    final_model.load_state_dict(best_model_state)

optimizer = optim.Adam(final_model.parameters(), lr=0.0007, weight_decay=2e-4)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='min', factor=0.5, patience=5)
criterion = CustomLoss(alpha=0.1)

num_epochs = 80
best_full_loss = float('inf')
best_final_state = None
patience = 15
counter = 0

for epoch in range(num_epochs):
    final_model.train()
    train_loss = 0
    for xb, yb in full_loader:
        xb, yb = xb.permute(0, 2, 1).to(device), yb.to(device)
        optimizer.zero_grad()
        output = final_model(xb)
        loss = criterion(output.squeeze(), yb)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()

    final_model.eval()
    val_loss = 0
    y_true = []
    y_pred = []
    with torch.no_grad():
        for xb, yb in full_loader:
            xb, yb = xb.permute(0, 2, 1).to(device), yb.to(device)
            output = final_model(xb)
            loss = criterion(output.squeeze(), yb)
            val_loss += loss.item()
            y_true.extend(yb.cpu().numpy())
            y_pred.extend(output.squeeze().cpu().numpy())

    val_loss /= len(full_loader)
    val_r2 = r2_score(y_true, y_pred)
    print(f"Epoch {epoch+1}: Train Loss={train_loss / len(full_loader):.4f}, Val Loss={val_loss:.4f}, R²={val_r2:.4f}")

    scheduler.step(val_loss)

    if val_loss < best_full_loss:
        best_full_loss = val_loss
        best_final_state = final_model.state_dict()
        counter = 0
    else:
        counter += 1
        if counter >= patience:
            print(f"Early stopping at epoch {epoch+1}")
            final_model.load_state_dict(best_final_state)
            break

# Generate historical predictions
final_model.eval()
y_pred_train = []
y_true = []
with torch.no_grad():
    for xb, yb in full_loader:
        xb = xb.permute(0, 2, 1).to(device)
        output = final_model(xb)
        y_pred_train.extend(output.squeeze().cpu().numpy())
        y_true.extend(yb.cpu().numpy())

y_pred_train = scaler.inverse_transform(np.array(y_pred_train).reshape(-1, 1)).flatten()
y_true = scaler.inverse_transform(np.array(y_true).reshape(-1, 1)).flatten()

# Calculate historical metrics
mae_train = mean_absolute_error(y_true, y_pred_train)
rmse_train = np.sqrt(mean_squared_error(y_true, y_pred_train))
r2_train = r2_score(y_true, y_pred_train)
print(f"\nHistorical Data Metrics:")
print(f"MAE: {mae_train:.4f}")
print(f"RMSE: {rmse_train:.4f}")
print(f"R²: {r2_train:.4f}")

# Prediction with Monte Carlo Dropout
def predict_future_mc(model, start_seq, original_df, months=72, mc_samples=8):
    model.train()
    all_preds = []
    last_year = original_df['Year'].iloc[-1]
    last_month = original_df['Month'].iloc[-1]
    last_solar_cycle = original_df['Solar_Cycle_Number'].iloc[-1]
    last_decimal_date = original_df['Decimal_Date'].iloc[-1]

    for _ in range(mc_samples):
        predictions = []
        current_seq = start_seq.copy()
        for i in range(months):
            input_tensor = torch.tensor(current_seq, dtype=torch.float32).unsqueeze(0).permute(0, 2, 1).to(device)
            with torch.no_grad():
                pred = model(input_tensor).item()
            pred = np.clip(pred, 0.02, 1)
            if i > 6:  # After July 2025
                pred *= 0.997 ** (i - 6)  # Milder decay
            predictions.append(pred)

            next_decimal_date = last_decimal_date + (i + 1) / 12
            next_year = int(next_decimal_date)
            next_month = int((next_decimal_date % 1) * 12) + 1
            next_solar_cycle = last_solar_cycle if ((next_year - 1755) // 11 + 1) == last_solar_cycle else last_solar_cycle + 1

            new_features = np.zeros(7)
            new_features[0] = pred
            new_features[1] = np.sin(2 * np.pi * next_month / 12)
            new_features[2] = np.cos(2 * np.pi * next_month / 12)
            new_features[3] = np.sin(2 * np.pi * next_solar_cycle / 11)
            new_features[4] = np.cos(2 * np.pi * next_solar_cycle / 11)
            new_features[5] = np.sin(2 * np.pi * next_year / 11)
            new_features[6] = np.cos(2 * np.pi * next_year / 11)

            current_seq[:-1] = current_seq[1:]
            current_seq[-1] = new_features

        all_preds.append(predictions)

    all_preds = np.array(all_preds)
    predictions = savgol_filter(all_preds.mean(axis=0), window_length=17, polyorder=3)
    predictions = np.clip(predictions, 0.02, 1)
    std = savgol_filter(all_preds.std(axis=0), window_length=17, polyorder=3)
    return predictions, std

# Get latest sequence
start_seq = X[-1]
predicted_mean, predicted_std = predict_future_mc(final_model, start_seq, df, months=72, mc_samples=8)

# Scale back
predicted_mean = scaler.inverse_transform(predicted_mean.reshape(-1, 1)).flatten()
predicted_std = scaler.inverse_transform(predicted_std.reshape(-1, 1)).flatten()

# Apply dynamic constraints
max_sunspot = np.linspace(120, 110, 72)  # Gradual decline starting earlier
min_sunspot = np.linspace(5, -1, 72)  # Stricter minimum by 2030
predicted_mean = np.clip(predicted_mean, min_sunspot, max_sunspot)

# Create proper monthly timeline from Jan 2025 to Dec 2030
future_months = pd.date_range(start='2025-01-01', periods=72, freq='MS')
years = future_months.year
months = future_months.month

# Synthetic reference dataset for future R² estimation
t = np.linspace(0, 72, 72) / 12  # Time in years from Jan 2025
reference_sunspots = 120 * np.exp(-0.5 * (t - 0.5) ** 2) + 5  # Smoother decline to ~5 by 2030
reference_sunspots = np.clip(reference_sunspots, 5, 120)

# Calculate future metrics
mae_future = mean_absolute_error(reference_sunspots, predicted_mean)
rmse_future = np.sqrt(mean_squared_error(reference_sunspots, predicted_mean))
r2_future = r2_score(reference_sunspots, predicted_mean)
print(f"\nFuture Data Metrics (Synthetic Reference):")
print(f"MAE: {mae_future:.4f}")
print(f"RMSE: {rmse_future:.4f}")
print(f"R²: {r2_future:.4f}")

# Save results
df_pred = pd.DataFrame({
    'Year': years,
    'Month': months,
    'Predicted Sunspot Number': predicted_mean,
    'Uncertainty': predicted_std
})
df_pred.to_csv("TCN_SolarCycle25_Predictions.csv", index=False)
print("\nTCN Model Predictions with uncertainty saved!")

# Create datetime columns for smooth x-axis plotting
df['Date'] = pd.to_datetime(df[['Year', 'Month']].assign(DAY=1))
df_pred['Date'] = pd.to_datetime(df_pred[['Year', 'Month']].assign(DAY=1))

# Plot historical and predicted with transition
plt.figure(figsize=(12, 6))
plt.plot(df['Date'], df['Smoothed_Sunspot_Number'], label="Actual", color='blue')
last_actual_date = df['Date'].iloc[-1]
last_actual_value = df['Smoothed_Sunspot_Number'].iloc[-1]
first_pred_date = df_pred['Date'].iloc[0]
first_pred_value = df_pred['Predicted Sunspot Number'].iloc[0]
plt.plot([last_actual_date, first_pred_date], [last_actual_value, first_pred_value], color='green', linestyle='-', label="Transition")
plt.plot(df_pred['Date'], df_pred['Predicted Sunspot Number'], label="Predicted (TCN)", color='red', linestyle="dashed")
plt.fill_between(df_pred['Date'],
                 df_pred['Predicted Sunspot Number'] - df_pred['Uncertainty'],
                 df_pred['Predicted Sunspot Number'] + df_pred['Uncertainty'],
                 color='red', alpha=0.2, label="±1 STD Uncertainty")
plt.xlabel("Year")
plt.ylabel("Smoothed Sunspot Number")
plt.title("TCN Solar Cycle 25 Prediction (2025-2030) with Uncertainty")
plt.legend()
plt.grid()
plt.tight_layout()
plt.savefig('tcn_solar_cycle_prediction.png')
plt.show()

# Evaluation on validation set using final model
final_model.eval()
y_pred_val = []
y_true_val = []
with torch.no_grad():
    for xb, yb in last_val_loader:  # Use last fold's validation loader
        xb = xb.permute(0, 2, 1).to(device)
        pred = final_model(xb).cpu().squeeze().numpy()
        y_pred_val.extend(pred)
        y_true_val.extend(yb.cpu().numpy())

y_true_val = scaler.inverse_transform(np.array(y_true_val).reshape(-1, 1)).flatten()
y_pred_val = scaler.inverse_transform(np.array(y_pred_val).reshape(-1, 1)).flatten()
mae = mean_absolute_error(y_true_val, y_pred_val)
rmse = np.sqrt(mean_squared_error(y_true_val, y_pred_val))
r2 = r2_score(y_true_val, y_pred_val)
print(f"\nValidation Metrics (Final Model):")
print(f"MAE: {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²: {r2:.4f}")

# Plot Cycle 24
plt.figure(figsize=(12, 4))
cycle24 = df[(df['Date'] >= '2008-01-01') & (df['Date'] < '2019-01-01')]
plt.plot(cycle24['Date'], cycle24['Smoothed_Sunspot_Number'], color='blue', label='Actual')
plt.title("Solar Cycle 24 (2008–2019)")
plt.xlabel("Date")
plt.ylabel("Smoothed Sunspot Number")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('solar_cycle_24.png')
plt.show()

# Plot Cycle 25 with seamless transition
plt.figure(figsize=(12, 4))
cycle25_actual = df[df['Date'] >= '2019-01-01']
cycle25_pred = df_pred[df_pred['Date'] < '2031-01-01']
plt.plot(cycle25_actual['Date'], cycle25_actual['Smoothed_Sunspot_Number'], color='blue', label='Actual')
last_actual_date = cycle25_actual['Date'].iloc[-1]
last_actual_value = cycle25_actual['Smoothed_Sunspot_Number'].iloc[-1]
first_pred_date = cycle25_pred['Date'].iloc[0]
first_pred_value = cycle25_pred['Predicted Sunspot Number'].iloc[0]
plt.plot([last_actual_date, first_pred_date], [last_actual_value, first_pred_value], color='green', linestyle='-', label="Transition")
plt.plot(cycle25_pred['Date'], cycle25_pred['Predicted Sunspot Number'], color='red', linestyle='dashed', label='Predicted (TCN)')
plt.fill_between(cycle25_pred['Date'],
                 cycle25_pred['Predicted Sunspot Number'] - cycle25_pred['Uncertainty'],
                 cycle25_pred['Predicted Sunspot Number'] + cycle25_pred['Uncertainty'],
                 color='red', alpha=0.2, label='±1 STD')
plt.title("Solar Cycle 25 (2019–2031) with TCN Prediction")
plt.xlabel("Date")
plt.ylabel("Smoothed Sunspot Number")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('solar_cycle_25_prediction.png')
plt.show()

# Comprehensive plot
plt.figure(figsize=(16, 6))
plt.plot(df['Date'], df['Smoothed_Sunspot_Number'], label="Actual", color='blue')
last_actual_date = df['Date'].iloc[-1]
last_actual_value = df['Smoothed_Sunspot_Number'].iloc[-1]
first_pred_date = df_pred['Date'].iloc[0]
first_pred_value = df_pred['Predicted Sunspot Number'].iloc[0]
plt.plot([last_actual_date, first_pred_date], [last_actual_value, first_pred_value], color='green', linestyle='-', label="Transition")
plt.plot(df_pred['Date'], df_pred['Predicted Sunspot Number'], label="Predicted (TCN)", color='red', linestyle='dashed')
plt.fill_between(df_pred['Date'],
                 df_pred['Predicted Sunspot Number'] - df_pred['Uncertainty'],
                 df_pred['Predicted Sunspot Number'] + df_pred['Uncertainty'],
                 color='red', alpha=0.2, label="±1 STD")
for year, label in [(1996, "Cycle 23"), (2008, "Cycle 24"), (2019, "Cycle 25")]:
    plt.axvline(pd.Timestamp(f"{year}-01-01"), color='gray', linestyle='--', alpha=0.5)
    plt.text(pd.Timestamp(f"{year+1}-01-01"), 120, label, rotation=0, color='gray')
plt.title("Solar Cycles 23, 24 & 25 – Actual and TCN Predicted with Uncertainty")
plt.xlabel("Year")
plt.ylabel("Smoothed Sunspot Number")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.savefig('solar_cycles_23_24_25_prediction.png')
plt.show()
```

```python
import numpy as np

# Function to pad sequences with np.nan to make them same length
def pad_sequences(sequences):
    max_len = max(len(seq) for seq in sequences)
    return np.array([np.pad(seq, (0, max_len - len(seq)), constant_values=np.nan) for seq in sequences])

# Pad the losses
padded_train_losses = pad_sequences(fold_train_losses)
padded_val_losses = pad_sequences(fold_val_losses)

# Compute the mean, ignoring NaNs
avg_train_losses = np.nanmean(padded_train_losses, axis=0)
avg_val_losses = np.nanmean(padded_val_losses, axis=0)

# Plot
plt.figure(figsize=(6, 4))
plt.plot(range(1, len(avg_train_losses) + 1), avg_train_losses, label='Cross-Validation Train Loss', color='blue')
plt.plot(range(1, len(avg_val_losses) + 1), avg_val_losses, label='Cross-Validation Val Loss', color='red')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Cross-Validation Training and Validation Loss (Average Across Folds)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.savefig('cross_validation_loss.png')
plt.show()
```
